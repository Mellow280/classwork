```python
# chess with lists

pawns = ['p'] * 8
cool = ['r', 'n', 'b', 'q', 'k', 'b', 'n', 'r']
empty = [' '] * 8

chess = [
    cool.copy(),
    pawns.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    empty.copy(),
    pawns.copy(),
    cool.copy(), 
]

chess
```

```python
chess[0][1] = ' '
chess[2][2] = 'n'
chess[5][6] = 'n'
chess
```

```python
# new data type

# tuples are immutable, lists are mutable

#not mutable (no changing)
# allows duplicates
# indexed ( has location)
tpl = (1, 2, 3, 'something')
tpl2 = tuple((1, 2, 3, 4, 5,))
print(tpl)
print(tpl2)
```

```python
tpl.index(3)
```

```python
# set

# can change (mutable)
# no duplicates
# unordered (no location)

set1 = {1, 2, 3, 1, 2, 3,} # the bad way
set2 = set([1,2,3,1,2,3,3,3,]) # the good way
set2 
```

```python
len(set2)
```

```python
set2.add(7)
len(set2)
```

```python
set2.remove(3)
len(set2)
```

```python
set2.difference(set1)
```

```python
set2 - set1
```

```python
# dictionary
# key: value pairs
# indexed by its key (not by its position) also referred to as string
# can change (mutable)
# keys cannot be duplicated, values can be duplicated

siren = {
    'name': 'siren',
    'age': 5,
    'species': 'rottweiler',
}
person = {
    'name': 'john',
    'age': 30,
    'city': 'charlotte',
    'wealth': 10,
    'animals': [
        siren,
    ],
}
person
```

```python
person['animals']
```

```python
person['city']
```
