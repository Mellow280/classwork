- correlation 
used for bivariate stats, looking at 2(numerical) columns from single observation(2 variables from single point)
measures the overlap between variables (how big is the overlap)
associated calculation 
demonstrate relationships between data

- Independant & Dependant variables
- - dependant 
    what you want to predict or explain
- - independant
    the variable that explains or predicts the outcome of the dependant variable 
Ex. age = independant
    wisdom = dependant 

- Pearson 'r' correlation
measures the strength and direction of teh linear relationships between two variables
ranges from -1 to 1
0 means no correlation 
1 means perfect positive correlation (as x increases y increases they move together)
-1 means perfect negative correlation(as x increase y decrease)
r sqaured = coefficient of determination 
cant have r without r squared (r**2) 

- Qualifications
- - must be continuous data
    not ranked, ordinal data, or categorial
- - must have linear relationship
    will not detect variability that expands or contracts at different point
- - Homoscedastic (equal variance)
    the variablilty of the dependant variable is the same at all levels of the independant variable 
    they move together 
skewness- indicates non linearity
kurt- indicate presence of outliears

- probability value
measures statistaical significance of the correlation, lower == better
p-value goes with r value
p-value <0.05 is typically considered significant 
p-value >0.05 is typically considered not significant
measures probability that future data will return 0.0

- 2 ways to calculate
- - scipy.stats.pearsonr(x,y) Ex. stats.pearsonr(df['x'],df['y'])
    includes p-value
- - df.corr(numeric_only=True)
    does not include p-value
    cheatcode to decide what calculations to check
