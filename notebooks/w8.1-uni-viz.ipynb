```python
import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
from scipy import stats # scipy is science python
# import scipy .stats
url = 'https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/soccer-players.csv'

df = pd.read_csv(url)
df.head()
```

```python
plt.figure(figsize=(3,3)) #outputs a jpg of our fixed image
plt.xlabel("day")
plt.ylabel("sales ($)")
plt.title("Fig 1: Daily coffee sales")
plt.xlim(0,7)
plt.ylim(0,60)
plt.grid(True)
txt = "Sales Item"
# plt.text(1, 20, txt, ha="center", va="center") 
plt.text(
    0.10,
    0.20,
    txt,
    ha="center", 
    color="blue",
    transform=plt.gca().transAxes
    )

plt.show() # plt is creating a new object
```

```python
# creating a histogram in plt

plt.figure(figsize=(3,3))
bins = df["height_in_cm"].count() ** (1/3) # ** means to the power of
plt.hist(df["height_in_cm"], bins=10, edgecolor="black")

plt.show()
```

```python
plt.figure(figsize=(5,3))
hist = sns.histplot(data=df, x="height_in_cm", kde=True)
plt.ylabel("frequency")
hist.set(xlabel="Height (cm)")
mean_val = df["height_in_cm"].mean()
median_val = df['height_in_cm'].median()
plt.axvline(
    mean_val,
    color='red',
    linestyle='dashed',
    linewidth=1,
    label='Mean'
    )
plt.axvline(
    median_val,
    color='blue',
    linestyle='--',
    label='Median'
)
skew = stats.skew(df['height_in_cm'], nan_policy='omit')
kurt = stats.kurtosis(df['height_in_cm'], nan_policy='omit')
txt = f"""skew: {skew:.2f} 
Kurt: {kurt:.2f}
"""
plt.text(
    0.05,
    0.6,
    txt,
    transform=plt.gca().transAxes,
)
hist.set(title='Histogram Soccer Player Heights')
sns.set_theme(style='darkgrid')
# .2f = skew to 2 decimals, f is a float
plt.legend()

plt.show()
```

```python
# plt boxplot
plt.figure(figsize=(8,5))
plt.boxplot(df['height_in_cm'].dropna(), orientation='horizontal')

plt.show()
```
