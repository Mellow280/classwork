```python
# string 
my_string = "this is my string."
print(my_string)
```

```python
my_string = "This is my new string."
print(my_string)
```

```python
print(my_string.upper())
print(my_string)
```

```python
# index of collection
print(len(my_string))
my_string[0] 
```

```python
my_string[-1]
```

```python
print(my_string)
ending = "this is goodbye"
print(ending)
my_string = my_string + ending
print(my_string)
```

```python
my_string: str = "this string is type hinted"
print(str(my_string))
```

```python
# integers 
a: int = 25 
b: int = 7
print('addition: ',a + b) # addition
print(f'division: {a/b}') # division
print(f'multiplication: {a*b}') # multiplication
print(f'subtraction: {a-b}') # subtraction
print(f'exponents: {a**b}') # exponentiation 
print(f'floor division: {a//b}') # floor division
print(f'modulus: {a % b}') # modulus
```

```python
a = str(a)
b = float(b)

print(f"new type a and b: {a.zfill(5),b}")
print(int(3.9999))
```

```python
# booleans 
x: bool = True
y: bool = False
print(f'x = {x}')
print(f'y = {y}')
print(f'b as bool: {bool(b)}')
print(f'0 as bool: {bool(0)}')
print(f'empty string as bool: {bool("")}')
print(f'a string as bool: {bool(a)}')
```

```python
# lists 
# ordered (index)
# mutable
# allows repeated values

my_list = [1, 2, '3', False, [1,2,3]]
print(my_list)
```

```python
my_list[-1]
list2 = list((1,2,3))
list2
print(list2)
print(list2.pop())
print(list2)
```

```python
l1 = []
l1.append(1)
print(l1)
l1.append(2)
print(l1)
l1.pop()
print(l1)
```
