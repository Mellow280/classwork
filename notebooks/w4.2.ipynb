```python
# Lord of Python
# a hero, mini boss, a final monster, using potion/items in battle, 3 different locations

from typing import TypedDict, List, Literal, Dict, Tuple
import random, time 

class CharacterTypes(TypedDict):
    name: str
    int_value: int
    attack_power: int
    defense: int
    strength: str 

strengths_weaknesses = {
    'fire': 'water',
    'water': 'earth',
    'earth': 'fire',
}

total_points_available = 20

character: CharacterTypes = {
    'name': 'Hero',
    'health': 90,
    'attack_power': 10,
    'defense': 10,
    'strength': '',
}  

enemy_list: List[CharacterTypes] = [
    {
        'name': 'Goblin',
        'health': 20,
        'attack_power': 5,
        'defense': 1,
        'strength': 'water'
    }
]

locations = ['Forest']

```

```python
for key in character:
    if key == 'name':
        name = input('what is your hero called?')
        character['name'] = name
        character['name']
    elif key == "strength":
        strength = input(f'what strength do you choose: {', '.join(strengths_weaknesses)}')
        if strength.lower() not in [k.lower() for k in strengths_weaknesses]:
            print(f'{strength} is not a possible option bozo!')
            continue
        character['strength'] = strength.lower() 
    else:
    # looped until the right input is put
        while True:
            int_value = input(f'You have {total_points_available} pts. How much do you want to add to {character['name']}s {key}')
            if not int_value.isnumeric():
                print('Hey, pass in an int, buddy!')
                continue
            int_value = int(int_value)
            if int_value > total_points_available:
                print(f'You dont have {int_value} points available. You have {total_points_available} points')
                continue 
            total_points_available -= int_value # tpa = tpa - int_value
            character[key] += int_value 
            print(f'You used {int_value} pts')
            print(f'congratulations,{character['name']} has {character[key]} {key} pts!')
            break
```

```python
# build the combat.
# the fight is a while loop
def attack(
    attacker: CharacterTypes, 
    defender: CharacterTypes
    ) -> Tuple[CharacterTypes]:
    """ 
    handles attack sequence
    
    Args: 
        attacker (CharacterTypes): the character doing the attacking 
        defender (CharacterTypes): the character being attacked
    Returns: 
        Tuple(CharacterTypes): (character, defender)

    """
    att_power = attacker['Attack_power'] * 2 if attacker['strentgh'] == strengths_weaknesses[defender['strength']] else attacker['attack_power']
    hit_prob = random.random()
    if hit_prob > 0.9: # miss 10% of time (90% accuracy)
        print(f'{attacker['name']}s attack missed.')
        return attacker, defender
    point_reduction = att_power - defender['defense']
    defender['heath'] -= point_reduction
    print(f'{attacker['name']} hit {defender['name']} for {point_reduction} pts.')
    return attacker, defender

for loc in locations:
    while True:
        hero_location = input(f'You must choose a location to go to: {', '.join(locations)}')
        hero_location = hero_location.capitalize()
        if hero_location not in locations:
            print(f'{hero_location} not possible')
            continue
        break
    hero_loc_index = locations.index(hero_location)
    enemy = enemy_list[hero_loc_index]
    print(f'In the {hero_location}, you met a {enemy} and it charges right at you!')
    actions = ['attack', 'wait']
    while character['health'] > 0 and enemy['health'] > 0:
        while True:
            action = input(f'what do you want to do? {", ".join(actions)}')
            action = action.lower()
            if action not in actions:
                print(f'{action} is not an option.')
                continue
            break
        if action == 'attack':
            print(f'{character['name']} chose to attack')
            character, enemy = attack(character, enemy)
        else:
            # enemy waits 20% of time
            print(f"{character['name']} chose to wait. It is now {enemy['name']}'s turn.")
            wait_prob = random.random()
            if wait_prob < 0.2:
                print(f'{enemy['name']} chose to wait as well.')
            else:
                enemy, character = attack(enemy, character)
        
        # enemy's turn
        enemy, character = attack(enemy, character)
    
    #after fight sequence complete
    locations.pop()
```
