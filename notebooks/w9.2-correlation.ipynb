```python
df['is_pass'] = (df['eventName'] == "Pass").astype(int)
df['is_pass']

df.head()
#true = 1
#false = 0
```

```python
df['is_shot'] = (df['eventName'] == "Shot").astype(int)
df.head()
```

```python
groupbys = [
    ['teamId', 'matchId'],
    ['teamId'],
    ['matchId']
]

for grp in groupbys:
    new_g = df.groupby(grp).agg(
        passes=('is_pass', 'sum'),
        shots=('is_shot', 'sum'),
        games=('matchId', 'nunique'),
    ).reset_index()
    r, p = stats.pearsonr(new_g['passes'], new_g['shots']) #when inputting into pearsonr it is Y before X
    r2 = r**2
    print(f'{grp}: ')
    print(f'\tr:{r:.4f}, p:{p:.4f}, r2:{r2:.4f}')
    if grp == ['teamId']:
        norm_new_g = new_g.copy()
        norm_new_g['pass_p_g'] = norm_new_g['passes'] / norm_new_g['games']
        norm_new_g['shot_p_g'] = norm_new_g['shots'] / norm_new_g['games']
        r1, p1 = stats.pearsonr(norm_new_g['pass_p_g'], norm_new_g['shot_p_g']) #when inputting into pearsonr it is Y before X
        r21 = r1**2
        print(f'{grp}: ')
        print(f'\tr:{r1:.4f}, p:{p1:.4f}, r2:{r21:.4f}')
    # print(grp, new_g)
```
