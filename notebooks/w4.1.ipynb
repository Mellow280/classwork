```python
painting1 = {
    'color': 'red',
    'size': '8x11'
}
if painting1['color'] == "blue":
    painting1['color'] = 'red'
elif painting1['color'] == 'black':
    painting1['color'] = 'blue'
elif painting1['color'] == 'red':
    painting1['color'] = 'black'
else:
    print('no change to color')
print(painting1)
```

```python
# Loops- continue to perform a set of code
# code block is ':' then an indentation 
counter = 0
while counter < 10:
    print(f'counter= {counter}')
    counter += 1 # counter = counter + 1
    
is_true = True

while is_true:
    print(is_true)
    is_true = False
```

```python
#for loop
# range(10) => 0...9, the number in() is exclusive
# range(0,10) => 0...9
# range(1,12) => 1...11

for temp_var in range(12):
    print(temp_var * 12)
```

```python
l1 = [1,2,3,4,5]
for num in l1:
    print(num)
```

```python
# generative loop
# varname = [temp_var we want for temp_var in var]
# the temporary variable 
l2 = [num + 12 for num in l1]
l2
```
