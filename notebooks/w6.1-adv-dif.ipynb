```python
df = pd.read_csv("../data/french-ambs.csv")
df.head()
```

```python
df = df.drop(columns=['lat_long']) # .drop exist on a dataframe
df.head()
```

```python

df = df[['name', 'place', 'year']]
df.head()
```

```python
df['place'].unique()
```

```python
df['place'].count()
df['place'].astype
```

```python
# Loops
#iterrows / itertuples
#iterrows
for idx, row in df.iterrows():
    if row['place'] == 'Holy Roman Empire':
        df.at[idx, 'place'] = 'HRE'
        
df['place'].unique()
```

```python
for row in df.itertuples():
    idx = row.Index
    if row.place == "Ottoman Empire":
        df.at[idx,'place'] = "Ott"
        
df['place'].unique()
```

```python
# using .loc to replace
df.loc[df['place'] == 'Netherlands', 'place'] = 'Neth'
df['place'].unique()
```

```python
#hash is dict
# replace many with a hash using .replace
hash = {place: f'{place[:3] + '.' if len(place) > 4 else place}' for place in df['place'].unique()}
df['place'] = df['place'].replace(hash)
df['place'].unique()
```

```python
# complex boolean logic with | (or), & (and), ~ (not)
df.loc[((df['place'] == 'Netherlands') | (df['place'] == 'England')) & (df['year'] <1550)]
```

```python
df.loc[~((df['place'] == 'Netherlands')) & (df['year'] >1588) & (df['year'] <1592)]
```
